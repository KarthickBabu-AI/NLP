```python
import pandas as pd
import numpy as np
import itertools
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, HashingVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import PassiveAggressiveClassifier
from sklearn.linear_model import LogisticRegression
from sklearn import metrics
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("news.csv")
df
```

```python
dataset=df.drop("Unnamed: 0",axis=1)
dataset
```

```python
y=dataset["label"]
```

```python
X_train,X_test,y_train,y_test=train_test_split(dataset['text'],y,test_size=0.33,random_state=53)
```

```python
HVec=HashingVectorizer(stop_words='english')
HVec_train=HVec.fit_transform(X_train)
HVec_test=HVec.transform(X_test)
#print(count_test)
```

```python
#dir(HVec)
```

```python
len(HVec.build_analyzer()(dataset["text"].iloc[0]))#no. of unique words
```

```python
#print(count_train.toarray())
```

```python
HV=LogisticRegression()
HV.fit(HVec_train,y_train)
pred=HV.predict(HVec_test)
score=metrics.accuracy_score(y_test,pred)
print("Accuary: %0.2f"%score)
cm = metrics.confusion_matrix(y_test, pred, labels=['FAKE', 'REAL'])
print("Confusion Matrics")
print("                Predicted")
print("               Fake  Real")
print("Actual Fake     " ,cm[0])
print("Actual real     " ,cm[1])
```

```python
from sklearn.metrics import classification_report as clf1
report = clf1(y_test,pred)
print(report)
```
